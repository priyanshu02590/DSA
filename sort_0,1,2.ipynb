```python
def sortColors(arr):
    mid=0
    high=len(arr)-1
    low=0
    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        else:  # nums[mid] == 2
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
```

```python
import numpy as np
arr=np.array([0,0,1,0,1,1,2,1,0,1,2,2,0,1])
```

```python
sortColors(arr)
print(arr)
```
