```python
import numpy as np
arr=np.array([7,4,3,5,1])
```

```python
def bubblesort(arr):
    n=len(arr)
    for i in range(n):
        c=0
        for j in range(n-i-1):
            if(arr[j]>arr[j+1]):
                temp=arr[j]
                arr[j]=arr[j+1]
                arr[j+1]=temp
                c+=1
            print(arr)
        if(c==0):
            break
```

```python
print(arr)
print("working")
bubblesort(arr)
```
