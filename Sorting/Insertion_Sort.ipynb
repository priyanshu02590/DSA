```python
import numpy as np
arr=np.array([7,4,3,5,1])
```

```python
def insertionsort(arr):
    n=len(arr)
    for i in range(1,n):
        key=arr[i]
        j=i-1
        while(j>=0 and arr[j]>key):
            arr[j+1]=arr[j]
            j-=1
        arr[j+1]=key
        print(arr)
```

```python
print(arr)
print("working")
insertionsort(arr)
```
