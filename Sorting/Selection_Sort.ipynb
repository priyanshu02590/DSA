```python
import numpy as np
arr=np.array([7,4,3,5,1])
```

```python
def selectionsort(arr):
    n=len(arr)
    for i in range(n):
        m=arr[i]
        ind=i
        for j in range(i+1,n):
            if(arr[j]<m):
                m=arr[j]
                ind=j
        temp=arr[i]
        arr[i]=arr[ind]
        arr[ind]=temp
        print(arr)
        
```

```python
print(arr)
print("working")
selectionsort(arr)
```
