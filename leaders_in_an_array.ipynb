```python
def leader_in_array(arr):
    n=len(arr)
    a=[]
    for i in range(n):
        c=0
        for j in range(i+1,n):
            if(arr[i]<arr[j]):
                c+=1
        if(c==0):
            a.append(arr[i])
    return a 
```

```python
import numpy as np
arr=np.array([10,22,12,3,0,6])
a=leader_in_array(arr)
print(a)
```

```python
def leader_in_array_2(arr):
    a=[]
    m=arr[-1]
    n=len(arr)
    for i in range(n-1,-1,-1):
        if(arr[i]>m):
            a.append(arr[i])
            m=arr[i]
    return a
```

```python
import numpy as np
arr=np.array([10,22,12,3,0,6])
a=leader_in_array(arr)
print(a)
```
