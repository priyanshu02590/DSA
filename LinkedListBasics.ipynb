```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
```

```python
a=Node(1)
b=Node(2)
c=Node(3)
a.next=b
b.next=c
print(a)
head=a
```

```python
def traverse(head):
    curr=head
    while(curr!=None):
        print(curr.data)
        curr=curr.next

traverse(a)
```

```python
def insert_at_beg(newnode,head):
    curr=head
    newnode.next=head
    return newnode
d=Node(4)
head=insert_at_beg(d,head)
traverse(d)
```

```python
def insert_at_end(newnode,head):
    curr=head
    while(curr.next!=None):
        curr=curr.next
    curr.next=newnode
e=Node(5)
insert_at_end(e,head)
traverse(d)
```

```python
def insert_at_kth(k,newnode,head):
    curr=head
    for i in range(k-2):
        curr=curr.next
    newnode.next=curr.next
    curr.next=newnode
e=Node(6)
insert_at_kth(3,e,head)
traverse(d)
```

```python
def delete_from_beg(head):
    return head.next
head=delete_from_beg(head)  
traverse(head)
```

```python
def delete_from_end(head):
    curr=head
    while(curr.next.next!=None):
        curr=curr.next
    curr.next=None
delete_from_end(head)
traverse(head)
```

```python
def delete_from_kth(k,head):
    curr=head
    for i in range(k-2):
        curr=curr.next
    curr.next=curr.next.next
delete_from_kth(3,head)
traverse(head)
```
