```python
import numpy as np
arr=np.array([1,3,6,7,9,11,12])
key=9
```

```python
def binary_search(arr,key):
    l=0
    h=len(arr)-1
    while(l<=h):
        mid=(l+h)//2
        if(arr[mid]==key):
            return mid
        if(arr[mid]<key):
            l=mid+1
        else:
            h=mid-1
    return -1
```

```python
a=binary_search(arr,key)
```

```python
print(a,arr[a])
```
