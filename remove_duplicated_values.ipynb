```python
import numpy as np
arr=np.array([1,1,23,9,4,5,7,7])
```

```python
def remove_duplicates(arr):
    d = {}

    for i in range(len(arr)):
        if arr[i] in d:
            d[arr[i]] += 1
        else:
            d[arr[i]] = 1
    new_arr = []

    for k, v in d.items():
        new_arr.append(k)

    return new_arr
```

```python
a=remove_duplicates(arr)
print(a)
```
