```python
def sort_alphabet(star):
    arr = [0] * len(star)
    for i in range(len(star)):
        arr[i] = star[i]
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] > arr[j]:
                arr[i], arr[j] = arr[j], arr[i]

    ans = ""

    for i in range(len(arr)):
        ans += arr[i]

    return ans


star = "heafeeufhoeh"

print(sort_alphabet(star))
```
