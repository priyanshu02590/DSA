```python
def union_two_arrays(arr1,arr2):
    d={}
    res=[]
    for i in range(len(arr1)):
        if(arr1[i] not in d):
            d[arr1[i]]=1
            res.append(arr1[i])
    for i in range(len(arr2)):
        if(arr2[i] not in d):    
            d[arr2[i]]=1
            res.append(arr2[i])
    return res
```

```python
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([2,3,4,5,])
a=union_two_arrays(arr1,arr2)
print(a)
```
